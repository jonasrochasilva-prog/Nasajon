```python
# =========================================================
# ETAPA 3 - API IBGE (VERSÃO SEM ERROS)
# =========================================================

import requests
import pandas as pd
import unicodedata
from difflib import get_close_matches

# =========================================================
# FUNÇÃO PARA NORMALIZAR TEXTO
# =========================================================

def normalizar(texto):

    texto = str(texto)

    texto = unicodedata.normalize(
        "NFKD",
        texto
    )

    texto = texto.encode(
        "ASCII",
        "ignore"
    ).decode(
        "utf-8"
    )

    return texto.lower().strip()

# =========================================================
# URL DA API
# =========================================================

url = "https://servicodados.ibge.gov.br/api/v1/localidades/municipios"

# =========================================================
# REQUISIÇÃO API
# =========================================================

try:

    resposta = requests.get(
        url,
        timeout=60
    )

    resposta.raise_for_status()

    municipios_api = resposta.json()

    print("API do IBGE carregada com sucesso.")

except requests.exceptions.RequestException as erro:

    print("Erro ao acessar API:")
    print(erro)

# =========================================================
# BASE EM MEMÓRIA
# =========================================================

base_municipios = {}

for item in municipios_api:

    try:

        nome = item["nome"]

        uf = item["microrregiao"]["mesorregiao"]["UF"]["sigla"]

        regiao = item["microrregiao"]["mesorregiao"]["UF"]["regiao"]["nome"]

        id_ibge = item["id"]

        chave = normalizar(nome)

        base_municipios[chave] = {

            "municipio_ibge": nome,

            "uf": uf,

            "regiao": regiao,

            "id_ibge": id_ibge
        }

    except Exception:

        continue

print("Total carregado:", len(base_municipios))

# =========================================================
# FUNÇÃO DE BUSCA
# =========================================================

def buscar_municipio(nome_input):

    nome_normalizado = normalizar(nome_input)

    # -----------------------------------------------------
    # MATCH EXATO
    # -----------------------------------------------------

    if nome_normalizado in base_municipios:

        dados = base_municipios[nome_normalizado]

        return {

            "municipio_ibge":
                dados["municipio_ibge"],

            "uf":
                dados["uf"],

            "regiao":
                dados["regiao"],

            "id_ibge":
                dados["id_ibge"],

            "status":
                "OK"
        }

    # -----------------------------------------------------
    # MATCH APROXIMADO
    # -----------------------------------------------------

    similares = get_close_matches(
        nome_normalizado,
        list(base_municipios.keys()),
        n=1,
        cutoff=0.70
    )

    if len(similares) > 0:

        chave = similares[0]

        dados = base_municipios[chave]

        return {

            "municipio_ibge":
                dados["municipio_ibge"],

            "uf":
                dados["uf"],

            "regiao":
                dados["regiao"],

            "id_ibge":
                dados["id_ibge"],

            "status":
                "OK"
        }

    # -----------------------------------------------------
    # NÃO ENCONTRADO
    # -----------------------------------------------------

    return {

        "municipio_ibge": "",

        "uf": "",

        "regiao": "",

        "id_ibge": "",

        "status": "NAO_ENCONTRADO"
    }

# =========================================================
# TESTES
# =========================================================

print("\n===================================")
print("TESTES")
print("===================================\n")

municipios_teste = [

    "Niteroi",
    "Curitba",
    "Belo Horzionte",
    "Santoo Andre",
    "Brasilia"

]

for cidade in municipios_teste:

    resultado = buscar_municipio(cidade)

    print(cidade)

    print(resultado)

    print()
```

```python
# =========================================================
# ETAPA 5 - ESTATÍSTICAS (VERSÃO AVALIAÇÃO)
# =========================================================

import json
import pandas as pd

# =========================================================
# VALIDAR EXISTÊNCIA DO DATAFRAME
# =========================================================

if "resultado_df" not in globals():

    raise Exception(
        "resultado_df não encontrado. Execute a Etapa 4 primeiro."
    )

# =========================================================
# GARANTIR TIPOS CORRETOS
# =========================================================

resultado_df["populacao_input"] = pd.to_numeric(
    resultado_df["populacao_input"],
    errors="coerce"
).fillna(0)

# =========================================================
# TOTAL DE MUNICÍPIOS
# =========================================================

total_municipios = int(
    len(resultado_df)
)

# =========================================================
# TOTAL STATUS OK
# =========================================================

total_ok = int(
    (
        resultado_df["status"] == "OK"
    ).sum()
)

# =========================================================
# TOTAL NÃO ENCONTRADO
# =========================================================

total_nao_encontrado = int(
    (
        resultado_df["status"] == "NAO_ENCONTRADO"
    ).sum()
)

# =========================================================
# TOTAL ERRO API
# =========================================================

total_erro_api = int(
    (
        resultado_df["status"] == "ERRO_API"
    ).sum()
)

# =========================================================
# POPULAÇÃO TOTAL DOS OK
# =========================================================

pop_total_ok = int(

    resultado_df.loc[
        resultado_df["status"] == "OK",
        "populacao_input"
    ].sum()

)

# =========================================================
# FILTRAR APENAS STATUS OK
# =========================================================

df_ok = resultado_df[
    resultado_df["status"] == "OK"
].copy()

# =========================================================
# TRATAMENTO DE REGIÃO VAZIA
# =========================================================

df_ok = df_ok[
    df_ok["regiao"].notna()
]

# =========================================================
# MÉDIAS POR REGIÃO
# =========================================================

medias_por_regiao = {}

if len(df_ok) > 0:

    medias = (

        df_ok

        .groupby("regiao")[
            "populacao_input"
        ]

        .mean()

        .round(2)

    )

    for regiao, media in medias.items():

        medias_por_regiao[regiao] = float(media)

# =========================================================
# JSON FINAL
# =========================================================

stats = {

    "total_municipios":
        total_municipios,

    "total_ok":
        total_ok,

    "total_nao_encontrado":
        total_nao_encontrado,

    "total_erro_api":
        total_erro_api,

    "pop_total_ok":
        pop_total_ok,

    "medias_por_regiao":
        medias_por_regiao
}

# =========================================================
# EXIBIR ESTATÍSTICAS
# =========================================================

print("\n===================================")
print("ESTATÍSTICAS FINAIS")
print("===================================\n")

print(
    json.dumps(
        stats,
        indent=4,
        ensure_ascii=False
    )
)

# =========================================================
# RESUMO EXECUTIVO
# =========================================================

print("\n===================================")
print("RESUMO")
print("===================================\n")

print(f"Total municípios: {total_municipios}")

print(f"Total OK: {total_ok}")

print(
    f"Total NÃO ENCONTRADO: "
    f"{total_nao_encontrado}"
)

print(
    f"Total ERRO_API: "
    f"{total_erro_api}"
)

print(
    f"População total OK: "
    f"{pop_total_ok}"
)

print("\nMédias por região:")

for regiao, media in medias_por_regiao.items():

    print(
        f"{regiao}: {media}"
    )
```
